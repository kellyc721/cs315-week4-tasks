## 4. Sentiment Analysis

The sentiment property returns a namedtuple of the form Sentiment(polarity, subjectivity). The polarity score is a float within the range [-1.0, 1.0]. The subjectivity is a float within the range [0.0, 1.0] where 0.0 is very objective and 1.0 is very subjective.

```python
testimonial = TextBlob("Textblob is amazingly simple to use. What great fun!")
testimonial.sentiment
```

```python
testimonial.sentiment.polarity
```

## 6. Words Inflection and Lemmatization

Each word in TextBlob.words or Sentence.words is a Word object (a subclass of unicode) with useful methods, e.g. for word inflection.

```python
sentence = TextBlob("Use 4 spaces per indentation level.")
sentence.words
```

```python
sentence.words[2].singularize()
```

```python
sentence.words[-1].pluralize()
```

Lemmatize words using .lemmatize() method:

```python
from textblob import Word
```

```python
w = Word("octopi")
w.lemmatize()
```

```python
w = Word("went")
w.lemmatize("v")  # Pass in WordNet part of speech (verb)
```

## 7. WordNet Integration

You can access the synsets for a Word via the synsets property or the get_synsets method, optionally passing in a part of speech.

```python
from textblob import Word
from textblob.wordnet import VERB
word = Word("octopus")
word.synsets
```

```python
Word("hack").get_synsets(pos=VERB)
```

You can access the definitions for each synset via the definitions property or the define() method:

```python
from textblob.wordnet import Synset
octopus = Synset("octopus.n.02")
shrimp = Synset("shrimp.n.03")
octopus.path_similarity(shrimp)
```

## 9. Spelling Correction

```python
b = TextBlob("I havv goood speling!")
print(b.correct())
```

## 10. Get Word and Noun Phrase Frequencies

We cab get the frequency of a word or noun phrase in a TextBlob by: 
1. word_counts dictionary
2. count() method

```python
monty = TextBlob("We are no longer the Knights who say Ni. "
                    "We are now the Knights who say Ekki ekki ekki PTANG.")
monty.word_counts['ekki']
```

word_counts is NOT case-sensitive; will return frequency of 0 if not found.

```python
monty.words.count('ekki')
```

For count(), you can specify whether or not the search should be case-sensitive (default is False).

```python
monty.words.count('ekki', case_sensitive=True)
```

Both can be used with nouns too:

```python
wiki.noun_phrases.count('python')
```

## 12. TextBlobs Are Like Python Strings!

You can use Python’s substring syntax for TextBlobs.

```python
zen[0:19]
```

```python
zen.upper()
```

```python
zen.find("Simple")
```

You can make comparisons between TextBlobs and strings.

```python
apple_blob = TextBlob("apples")
banana_blob = TextBlob("bananas")
apple_blob < banana_blob
```

```python
apple_blob == "apples"
```

You can concatenate and interpolate TextBlobs and strings.

```python
apple_blob + " and " + banana_blob
```

```python
"{0} and {1}".format(apple_blob, banana_blob)
```

## 14. Get Start and End Indices of Sentences

Use sentence.start and sentence.end to get the indices where a sentence starts and ends within a TextBlob.
